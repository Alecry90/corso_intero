## LOOPS METHODS ##
RANGE è un metodo utile per costruire sequenze numeriche iterabili per cicli
1. range= a[0=0], n, stride= 1

```python
for x in range(5):
    print(x)
```

```python
for x in range(1, 5):
    print(x)
```

```python
for x in range (0, 6, 2):
    print(x)        #da 0 a 6 con passo di 2
```

```python
for x in range(10, 0, -2):
    print(x)   #da 10 a 0 con passo di 2("-2")
```

1. INIZIO 
2. CONDIZIONE: IL METEO DI OGGI
3. STA PIOVENDO?

. NO: NON LO PRENDO
4. FINE
IF COND. POSITIVA= TRUE (prendo l'ombrello)
ELSE COND. NEGATIVA= FALSE (non lo prendo)
