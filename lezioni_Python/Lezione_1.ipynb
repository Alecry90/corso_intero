# STRINGA

## SONO UNA SEQUENZA ALFANUMERICA ORDINATA E IMMUTABILE
1. Ogni carattere all'interno della stringa ha un indice(ORDINATA)
2. GLI ELEMENTI SLL'INTERNO DELLA STRINGA NON LI POSSO CAMBIARE TRAMITE ASSEGNAZIONE


```python
a= "cat"   # oppure a= 'cat'
print(a)
# devo considerare che le lettere siano come comprese in una tabella e posso andare a richiamare le singole lettere
print(a[0])    #partono da 0
print(a[1])
print(a[2])
```

```python
#voglio trasformare la c da minuscolo a maiuscolo
a= "cat"  
print(a)
t = "C"  + a[1:] #sostituisce la prima lettera
print(t)
```

```python
b="elefante"
print(b[3])
v= "maial"+ b[2:]
print(v)
```

```python
f="superforte"
#VOGLIO ACCEDERE AL PENULTIMO ELEMENTO... PARTO DA -1 A -n
j=  f[:-2]+ 'se'
print(j)
```
