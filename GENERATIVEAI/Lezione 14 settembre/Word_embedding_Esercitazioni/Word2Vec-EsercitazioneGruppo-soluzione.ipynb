```python

# ============================================================
# ESERCIZIO COMPLETO:
# COSTRUIAMO UN SISTEMA PERSONALIZZATO DI EMBEDDINGS
# ============================================================

"""
OBIETTIVO DELL'ESERCIZIO

In questo esercizio costruiamo un piccolo sistema di ricerca
semantica basato su diversi metodi di rappresentazione del testo.

Confronteremo:

1. TF-IDF
2. Word2Vec
3. Una combinazione personalizzata di TF-IDF e Word2Vec

Successivamente miglioreremo il sistema aggiungendo:

- pesi diversi per le parole importanti;
- peso legato alla posizione delle parole;
- normalizzazione per la lunghezza dei documenti;
- filtri per categoria e data;
- metriche di valutazione;
- un semplice sistema di feedback;
- gestione di documenti in più lingue.

IMPORTANTE

Il modello Word2Vec utilizzato in questo esercizio viene addestrato
su un piccolo corpus didattico.

Per questo motivo i risultati non devono essere considerati
paragonabili a quelli ottenuti con modelli addestrati su milioni
di documenti.

L'obiettivo principale è capire il funzionamento del sistema.
"""


# ============================================================
# 1. IMPORTIAMO LE LIBRERIE
# ============================================================

import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from gensim.models import Word2Vec

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.decomposition import PCA


# ============================================================
# 2. DEFINIAMO I DOCUMENTI DI ADDESTRAMENTO
# ============================================================

"""
Word2Vec non lavora direttamente con frasi sotto forma di stringhe.

Ha bisogno di una lista di frasi, dove ogni frase è a sua volta
una lista di parole.

Esempio:

"Il gatto dorme"

diventa:

["il", "gatto", "dorme"]

Per questo motivo prepariamo un piccolo corpus tokenizzato.
"""

corpus_tokenizzato = [

    ["il", "machine", "learning", "è", "una", "branca",
     "dell", "intelligenza", "artificiale"],

    ["il", "deep", "learning", "utilizza", "reti", "neurali"],

    ["gli", "algoritmi", "di", "machine", "learning",
     "imparano", "dai", "dati"],

    ["la", "programmazione", "python", "è", "utile",
     "per", "la", "data", "science"],

    ["python", "viene", "utilizzato", "nella", "data",
     "science", "e", "nel", "machine", "learning"],

    ["l", "intelligenza", "artificiale", "permette",
     "di", "creare", "sistemi", "intelligenti"],

    ["le", "reti", "neurali", "sono", "utilizzate",
     "nel", "deep", "learning"],

    ["i", "gatti", "sono", "animali", "domestici",
     "molto", "indipendenti"],

    ["i", "cani", "sono", "animali", "domestici",
     "fedeli", "e", "giocherelloni"],

    ["gli", "animali", "domestici", "vivono",
     "spesso", "insieme", "alle", "persone"],

    ["la", "programmazione", "permette", "di",
     "sviluppare", "applicazioni", "informatiche"],

    ["la", "data", "science", "analizza", "grandi",
     "quantità", "di", "dati"]
]


# ============================================================
# 3. ADDESTRIAMO IL MODELLO WORD2VEC
# ============================================================

"""
Word2Vec impara un vettore numerico per ogni parola del vocabolario.

Parole che compaiono in contesti simili possono ottenere vettori
semanticamente vicini.

Parametri principali:

- vector_size=50:
  ogni parola viene rappresentata con un vettore di 50 numeri;

- window=3:
  il modello osserva le parole vicine entro una finestra di 3 parole;

- min_count=1:
  anche le parole che compaiono una sola volta vengono mantenute;

- workers=1:
  utilizziamo un solo processo per rendere il risultato più riproducibile;

- sg=1:
  utilizziamo il modello Skip-Gram.
"""

print("=" * 70)
print("1. ADDESTRAMENTO DEL MODELLO WORD2VEC")
print("=" * 70)

model_w2v = Word2Vec(
    sentences=corpus_tokenizzato,
    vector_size=50,
    window=3,
    min_count=1,
    workers=1,
    sg=1,
    epochs=200,
    seed=42
)

print("Modello Word2Vec addestrato correttamente.")
print("Dimensione dei vettori:", model_w2v.vector_size)
print("Numero di parole nel vocabolario:", len(model_w2v.wv))


# ============================================================
# 4. ESPLORIAMO GLI EMBEDDINGS DELLE PAROLE
# ============================================================

"""
Ogni parola presente nel vocabolario possiede un vettore.

Per esempio:

model_w2v.wv["python"]

restituisce un array di numeri.

Il vettore rappresenta la parola nello spazio degli embeddings.
"""

print("\n" + "=" * 70)
print("2. ESPLORAZIONE DEGLI EMBEDDINGS")
print("=" * 70)

parola_esempio = "python"

if parola_esempio in model_w2v.wv:

    vettore_python = model_w2v.wv[parola_esempio]

    print("Parola:", parola_esempio)
    print("Embedding della parola:")
    print(vettore_python)

    print("\nDimensione dell'embedding:", len(vettore_python))

else:

    print("La parola non è presente nel vocabolario.")


# ============================================================
# 5. CONFRONTIAMO PAROLE SIMILI E POCO SIMILI
# ============================================================

"""
La funzione similarity() calcola la similarità coseno tra
due vettori Word2Vec.

Il valore è generalmente compreso tra -1 e 1.

Un valore alto indica che i vettori hanno una direzione simile.

ATTENZIONE

Con un corpus molto piccolo i risultati possono essere poco
affidabili. In questo esercizio ci interessa soprattutto
comprendere il funzionamento della tecnica.
"""

print("\n" + "=" * 70)
print("3. SIMILARITÀ TRA PAROLE")
print("=" * 70)

coppie_parole = [
    ("machine", "learning"),
    ("python", "science"),
    ("gatti", "cani"),
    ("gatti", "python")
]

for parola1, parola2 in coppie_parole:

    if parola1 in model_w2v.wv and parola2 in model_w2v.wv:

        similarita = model_w2v.wv.similarity(
            parola1,
            parola2
        )

        print(
            f"Similarità tra '{parola1}' e '{parola2}': "
            f"{similarita:.4f}"
        )

    else:

        print(
            f"Una delle parole '{parola1}' o '{parola2}' "
            "non è presente nel vocabolario."
        )


# ============================================================
# 6. VISUALIZZIAMO GLI EMBEDDINGS DELLE PAROLE CON PCA
# ============================================================

"""
Gli embeddings Word2Vec hanno 50 dimensioni.

Non possiamo rappresentarli direttamente su un grafico 2D.

Utilizziamo quindi PCA, che riduce le 50 dimensioni a 2 dimensioni.

PCA cerca le direzioni che conservano la maggior parte
della variabilità dei dati.

Il grafico è soltanto una rappresentazione visiva approssimata.
"""

print("\n" + "=" * 70)
print("4. VISUALIZZAZIONE DEGLI EMBEDDINGS CON PCA")
print("=" * 70)

parole_da_visualizzare = [
    parola
    for parola in [
        "machine",
        "learning",
        "deep",
        "python",
        "science",
        "intelligenza",
        "artificiale",
        "reti",
        "neurali",
        "gatti",
        "cani",
        "animali"
    ]
    if parola in model_w2v.wv
]

vettori_parole = np.array([
    model_w2v.wv[parola]
    for parola in parole_da_visualizzare
])

if len(vettori_parole) >= 2:

    pca_parole = PCA(n_components=2)

    coordinate_parole = pca_parole.fit_transform(
        vettori_parole
    )

    plt.figure(figsize=(10, 7))

    plt.scatter(
        coordinate_parole[:, 0],
        coordinate_parole[:, 1]
    )

    for indice, parola in enumerate(parole_da_visualizzare):

        plt.annotate(
            parola,
            (
                coordinate_parole[indice, 0],
                coordinate_parole[indice, 1]
            )
        )

    plt.title("Word2Vec: visualizzazione degli embeddings con PCA")
    plt.xlabel("Componente principale 1")
    plt.ylabel("Componente principale 2")
    plt.grid(True)
    plt.show()


# ============================================================
# 7. FUNZIONE DI TOKENIZZAZIONE
# ============================================================

"""
La funzione seguente serve a trasformare un documento testuale
in una lista di parole.

Esempio:

"Il Machine Learning è utile!"

diventa:

["il", "machine", "learning", "è", "utile"]

Convertiamo tutto in minuscolo e rimuoviamo la punteggiatura.
"""

def tokenizza_documento(documento):

    documento = documento.lower()

    # Manteniamo lettere, numeri e caratteri accentati.
    token = re.findall(
        r"\b[\wàèéìòù]+\b",
        documento,
        flags=re.UNICODE
    )

    return token


# ============================================================
# 8. FUNZIONE DOCUMENTO_TO_W2V
# ============================================================

"""
Word2Vec produce embeddings per le singole parole.

Un documento, invece, contiene molte parole.

Per ottenere un unico vettore per tutto il documento,
calcoliamo la media degli embeddings delle parole presenti.

Esempio:

documento = ["python", "è", "utile"]

embedding_documento =
    media(
        embedding_python,
        embedding_è,
        embedding_utile
    )

Le parole sconosciute vengono ignorate.

Se nessuna parola del documento è presente nel vocabolario,
restituiamo un vettore di zeri.
"""

def documento_to_w2v(documento, model_w2v):

    # Se riceviamo una stringa, la tokenizziamo.
    if isinstance(documento, str):

        parole = tokenizza_documento(documento)

    else:

        parole = documento

    vettori_validi = []

    for parola in parole:

        if parola in model_w2v.wv:

            vettore = model_w2v.wv[parola]

            vettori_validi.append(vettore)

    # Se non troviamo parole conosciute,
    # restituiamo un vettore di zeri.
    if len(vettori_validi) == 0:

        return np.zeros(
            model_w2v.vector_size
        )

    # La media trasforma molti vettori di parole
    # in un unico vettore rappresentativo del documento.
    embedding_documento = np.mean(
        vettori_validi,
        axis=0
    )

    return embedding_documento


# ============================================================
# 9. DEFINIAMO I DOCUMENTI DI TEST
# ============================================================

"""
Questi sono i documenti sui quali testeremo il sistema.

Ogni documento rappresenta un possibile contenuto
che un utente potrebbe cercare.
"""

docs_test = [

    "Il machine learning è una branca dell'intelligenza artificiale",

    "I gatti sono animali domestici molto indipendenti",

    "La programmazione Python è utile per data science",

    "Gli algoritmi di deep learning usano reti neurali",

    "I cani sono compagni fedeli e giocherelloni"
]


query_test = "Voglio imparare l'intelligenza artificiale"


# ============================================================
# 10. CALCOLIAMO GLI EMBEDDINGS DEI DOCUMENTI CON WORD2VEC
# ============================================================

"""
Per ogni documento calcoliamo un vettore Word2Vec.

Alla fine otteniamo una matrice:

numero_documenti x dimensione_embedding

Nel nostro caso:

5 documenti x 50 dimensioni
"""

print("\n" + "=" * 70)
print("5. EMBEDDINGS WORD2VEC DEI DOCUMENTI")
print("=" * 70)

docs_w2v = np.array([
    documento_to_w2v(
        documento,
        model_w2v
    )
    for documento in docs_test
])

print("Forma della matrice degli embeddings:")
print(docs_w2v.shape)


# ============================================================
# 11. CALCOLIAMO GLI EMBEDDINGS DELLA QUERY
# ============================================================

query_w2v = documento_to_w2v(
    query_test,
    model_w2v
)

print("\nEmbedding della query calcolato.")
print("Dimensione embedding query:", len(query_w2v))


# ============================================================
# 12. RICERCA CON WORD2VEC
# ============================================================

"""
Confrontiamo l'embedding della query con gli embeddings
di tutti i documenti.

Utilizziamo la similarità coseno.

La similarità coseno misura quanto due vettori puntano
nella stessa direzione.

Un valore alto indica una maggiore somiglianza vettoriale.
"""

print("\n" + "=" * 70)
print("6. RICERCA CON WORD2VEC")
print("=" * 70)

similarita_w2v = cosine_similarity(
    [query_w2v],
    docs_w2v
)[0]

risultati_w2v = [

    (
        indice + 1,
        similarita,
        documento
    )

    for indice, (similarita, documento)
    in enumerate(
        zip(
            similarita_w2v,
            docs_test
        )
    )
]

risultati_w2v.sort(
    key=lambda elemento: elemento[1],
    reverse=True
)

print("\nTop 5 risultati Word2Vec:")

for posizione, (doc_id, sim, documento) in enumerate(
    risultati_w2v[:5],
    1
):

    print(
        f"{posizione}. Doc {doc_id} "
        f"(similarità: {sim:.4f}) - {documento}"
    )


# ============================================================
# 13. RICERCA CON TF-IDF
# ============================================================

"""
TF-IDF rappresenta i documenti con vettori basati
sulla frequenza delle parole.

TF-IDF assegna un peso maggiore alle parole:

- frequenti in un documento;
- poco frequenti negli altri documenti.

A differenza di Word2Vec, TF-IDF non comprende direttamente
il significato semantico delle parole.

Per esempio, "cane" e "animale" sono parole diverse
e vengono trattate come elementi distinti del vocabolario.
"""

print("\n" + "=" * 70)
print("7. RICERCA CON TF-IDF")
print("=" * 70)

tfidf_vectorizer = TfidfVectorizer(
    lowercase=True
)

matrice_tfidf = tfidf_vectorizer.fit_transform(
    docs_test
)

query_tfidf = tfidf_vectorizer.transform(
    [query_test]
)

similarita_tfidf = cosine_similarity(
    query_tfidf,
    matrice_tfidf
)[0]

risultati_tfidf = [

    (
        indice + 1,
        similarita,
        documento
    )

    for indice, (similarita, documento)
    in enumerate(
        zip(
            similarita_tfidf,
            docs_test
        )
    )
]

risultati_tfidf.sort(
    key=lambda elemento: elemento[1],
    reverse=True
)

print("\nTop 5 risultati TF-IDF:")

for posizione, (doc_id, sim, documento) in enumerate(
    risultati_tfidf[:5],
    1
):

    print(
        f"{posizione}. Doc {doc_id} "
        f"(similarità: {sim:.4f}) - {documento}"
    )


# ============================================================
# 14. CLASSE PERSONALIZZATA MIO SISTEMA EMBEDDINGS
# ============================================================

"""
La classe MioSistemaEmbeddings permette di costruire
un sistema di ricerca personalizzato.

Il parametro metodo può assumere diversi valori:

- "media":
  utilizza la media semplice degli embeddings Word2Vec;

- "pesato":
  utilizza una media pesata degli embeddings Word2Vec;

- "combinato":
  combina Word2Vec e TF-IDF;

- "posizione":
  utilizza anche un peso legato alla posizione della parola;

- "normalizzato":
  normalizza il documento in base alla sua lunghezza.

L'obiettivo non è creare un sistema industriale,
ma capire come si possono combinare diverse informazioni.
"""

class MioSistemaEmbeddings:

    def __init__(
        self,
        metodo="combinato",
        modello_w2v=None
    ):

        """
        Inizializza il sistema.

        Parametri
        ---------

        metodo:
            metodo di costruzione dell'embedding del documento.

        modello_w2v:
            modello Word2Vec già addestrato.
        """

        self.metodo = metodo

        self.modello_w2v = modello_w2v

        # Il vocabolario TF-IDF verrà creato durante l'addestramento.
        self.tfidf_vectorizer = None

        # Matrice TF-IDF dei documenti.
        self.matrice_tfidf = None

        # Matrice Word2Vec dei documenti.
        self.matrice_w2v = None

        # Matrice degli embeddings combinati.
        self.matrice_embeddings = None

        # Lista dei documenti utilizzati per l'addestramento.
        self.documenti = []

        # Dizionario che assegna un peso alle parole importanti.
        self.pesi_parole = {

            "machine": 2.0,
            "learning": 2.0,
            "deep": 2.0,
            "python": 1.5,
            "science": 1.5,
            "intelligenza": 2.0,
            "artificiale": 2.0,
            "neurali": 1.5,
            "reti": 1.5
        }


    # ========================================================
    # METODO DI SUPPORTO: EMBEDDING WORD2VEC DI UN DOCUMENTO
    # ========================================================

    def embedding_w2v_documento(
        self,
        documento,
        usa_pesi=False,
        usa_posizione=False
    ):

        """
        Calcola l'embedding Word2Vec di un documento.

        Possiamo utilizzare tre strategie:

        1. media semplice dei vettori;

        2. media pesata in base all'importanza delle parole;

        3. media pesata in base alla posizione delle parole.

        La posizione viene utilizzata in modo semplice:

        - le parole iniziali ricevono un peso leggermente maggiore;
        - le parole finali ricevono un peso leggermente minore.

        Questo non rappresenta una vera comprensione sintattica,
        ma serve a mostrare come si può introdurre informazione
        aggiuntiva nel calcolo dell'embedding.
        """

        parole = tokenizza_documento(documento)

        vettori = []
        pesi = []

        lunghezza_documento = len(parole)

        for posizione, parola in enumerate(parole):

            if parola not in self.modello_w2v.wv:

                # Se la parola non è nel vocabolario Word2Vec,
                # la ignoriamo.
                continue

            vettore = self.modello_w2v.wv[parola]

            # Ogni parola parte con peso 1.
            peso = 1.0

            # ------------------------------------------------
            # SFIDA 2:
            # AGGIUNGERE PESI DIVERSI PER PAROLE IMPORTANTI
            # ------------------------------------------------

            if usa_pesi:

                peso_parola = self.pesi_parole.get(
                    parola,
                    1.0
                )

                peso = peso * peso_parola

            # ------------------------------------------------
            # SFIDA 3:
            # CONSIDERARE LA POSIZIONE DELLE PAROLE
            # ------------------------------------------------

            if usa_posizione and lunghezza_documento > 0:

                posizione_relativa = posizione / lunghezza_documento

                # Le parole all'inizio ricevono un peso maggiore.
                # Il peso diminuisce gradualmente verso la fine.
                peso_posizione = 1.2 - (
                    0.4 * posizione_relativa
                )

                peso = peso * peso_posizione

            vettori.append(vettore)
            pesi.append(peso)

        # Se non ci sono parole conosciute,
        # restituiamo un vettore di zeri.
        if len(vettori) == 0:

            return np.zeros(
                self.modello_w2v.vector_size
            )

        vettori = np.array(vettori)
        pesi = np.array(pesi)

        # Media semplice degli embeddings.
        if not usa_pesi and not usa_posizione:

            embedding = np.mean(
                vettori,
                axis=0
            )

        else:

            # Media pesata degli embeddings.
            embedding = np.average(
                vettori,
                axis=0,
                weights=pesi
            )

        # ----------------------------------------------------
        # SFIDA 4:
        # NORMALIZZARE IN BASE ALLA LUNGHEZZA DEL DOCUMENTO
        # ----------------------------------------------------

        if self.metodo == "normalizzato":

            # Dividiamo per la radice quadrata della lunghezza.
            # In questo modo i documenti molto lunghi non
            # dominano automaticamente quelli più brevi.
            fattore_normalizzazione = np.sqrt(
                max(lunghezza_documento, 1)
            )

            embedding = embedding / fattore_normalizzazione

        return embedding


    # ========================================================
    # METODO DI ADDESTRAMENTO
    # ========================================================

    def addestra(self, documenti):

        """
        Addestra il sistema sui documenti ricevuti.

        Durante l'addestramento:

        1. salviamo i documenti;
        2. calcoliamo gli embeddings Word2Vec;
        3. costruiamo la matrice TF-IDF;
        4. combiniamo le rappresentazioni se richiesto.
        """

        print(
            f"\nAddestramento del sistema con metodo: "
            f"{self.metodo}"
        )

        self.documenti = documenti

        # ----------------------------------------------------
        # CREAZIONE DEGLI EMBEDDINGS WORD2VEC
        # ----------------------------------------------------

        usa_pesi = self.metodo in [
            "pesato",
            "combinato",
            "posizione",
            "normalizzato"
        ]

        usa_posizione = self.metodo in [
            "posizione",
            "combinato",
            "normalizzato"
        ]

        self.matrice_w2v = np.array([

            self.embedding_w2v_documento(
                documento,
                usa_pesi=usa_pesi,
                usa_posizione=usa_posizione
            )

            for documento in documenti
        ])

        # ----------------------------------------------------
        # CREAZIONE DELLA MATRICE TF-IDF
        # ----------------------------------------------------

        self.tfidf_vectorizer = TfidfVectorizer(
            lowercase=True
        )

        self.matrice_tfidf = self.tfidf_vectorizer.fit_transform(
            documenti
        )

        # ----------------------------------------------------
        # CREAZIONE DELL'EMBEDDING COMBINATO
        # ----------------------------------------------------

        if self.metodo == "combinato":

            """
            TF-IDF e Word2Vec hanno dimensioni diverse.

            Esempio:

            Word2Vec:
                50 dimensioni

            TF-IDF:
                una dimensione per ogni parola del vocabolario

            Non possiamo sommarli direttamente.

            Per ottenere un vettore combinato:

            1. calcoliamo il vettore Word2Vec;
            2. calcoliamo il vettore TF-IDF;
            3. riduciamo il vettore TF-IDF a 50 dimensioni
               utilizzando PCA;
            4. combiniamo i due vettori.

            In questo modo otteniamo un embedding finale
            di dimensione 50.
            """

            tfidf_dense = self.matrice_tfidf.toarray()

            dimensione_pca = min(
                50,
                tfidf_dense.shape[0],
                tfidf_dense.shape[1]
            )

            # Se la matrice è troppo piccola per PCA,
            # utilizziamo direttamente gli embeddings Word2Vec.
            if dimensione_pca >= 1:

                pca_tfidf = PCA(
                    n_components=dimensione_pca
                )

                tfidf_ridotto = pca_tfidf.fit_transform(
                    tfidf_dense
                )

                # Completiamo con zeri fino a 50 dimensioni.
                tfidf_completo = np.zeros(
                    (
                        len(documenti),
                        self.modello_w2v.vector_size
                    )
                )

                tfidf_completo[
                    :,
                    :dimensione_pca
                ] = tfidf_ridotto

                # Normalizziamo i due blocchi prima della combinazione.
                norma_w2v = np.linalg.norm(
                    self.matrice_w2v,
                    axis=1,
                    keepdims=True
                )

                norma_tfidf = np.linalg.norm(
                    tfidf_completo,
                    axis=1,
                    keepdims=True
                )

                norma_w2v[norma_w2v == 0] = 1
                norma_tfidf[norma_tfidf == 0] = 1

                w2v_normalizzato = (
                    self.matrice_w2v / norma_w2v
                )

                tfidf_normalizzato = (
                    tfidf_completo / norma_tfidf
                )

                # Combinazione dei due metodi.
                # Il peso 0.7 privilegia Word2Vec.
                self.matrice_embeddings = (
                    0.7 * w2v_normalizzato
                    +
                    0.3 * tfidf_normalizzato
                )

            else:

                self.matrice_embeddings = self.matrice_w2v

        else:

            # Per gli altri metodi utilizziamo gli embeddings Word2Vec.
            self.matrice_embeddings = self.matrice_w2v

        print("Addestramento completato.")

        print(
            "Forma matrice embeddings:",
            self.matrice_embeddings.shape
        )


    # ========================================================
    # METODO PER CONVERTIRE UN DOCUMENTO IN EMBEDDING
    # ========================================================

    def documento_to_embedding(self, documento):

        """
        Trasforma un documento in un vettore numerico.

        Se il metodo è combinato:

        - calcoliamo l'embedding Word2Vec;
        - calcoliamo il vettore TF-IDF;
        - trasformiamo il TF-IDF in una rappresentazione
          compatibile con Word2Vec;
        - combiniamo i due vettori.

        Negli altri casi utilizziamo l'embedding Word2Vec.
        """

        embedding_w2v = self.embedding_w2v_documento(
            documento,
            usa_pesi=self.metodo in [
                "pesato",
                "combinato",
                "posizione",
                "normalizzato"
            ],
            usa_posizione=self.metodo in [
                "posizione",
                "combinato",
                "normalizzato"
            ]
        )

        if self.metodo != "combinato":

            return embedding_w2v

        # ----------------------------------------------------
        # EMBEDDING TF-IDF DELLA QUERY
        # ----------------------------------------------------

        query_tfidf = self.tfidf_vectorizer.transform(
            [documento]
        ).toarray()

        dimensione_pca = min(
            50,
            self.matrice_tfidf.shape[0],
            self.matrice_tfidf.shape[1]
        )

        if dimensione_pca < 1:

            return embedding_w2v

        # Ricreiamo la riduzione PCA utilizzando
        # la matrice TF-IDF dei documenti.
        tfidf_documenti_dense = self.matrice_tfidf.toarray()

        pca_tfidf = PCA(
            n_components=dimensione_pca
        )

        pca_tfidf.fit(
            tfidf_documenti_dense
        )

        query_tfidf_ridotta = pca_tfidf.transform(
            query_tfidf
        )

        query_tfidf_completa = np.zeros(
            self.modello_w2v.vector_size
        )

        query_tfidf_completa[
            :dimensione_pca
        ] = query_tfidf_ridotta[0]

        # Normalizziamo i due vettori.
        norma_w2v = np.linalg.norm(
            embedding_w2v
        )

        norma_tfidf = np.linalg.norm(
            query_tfidf_completa
        )

        if norma_w2v != 0:

            embedding_w2v = embedding_w2v / norma_w2v

        if norma_tfidf != 0:

            query_tfidf_completa = (
                query_tfidf_completa / norma_tfidf
            )

        embedding_combinato = (
            0.7 * embedding_w2v
            +
            0.3 * query_tfidf_completa
        )

        return embedding_combinato


    # ========================================================
    # METODO PER TROVARE I DOCUMENTI SIMILI
    # ========================================================

    def trova_simili(
        self,
        query,
        documenti=None,
        n=3
    ):

        """
        Cerca i documenti più simili alla query.

        Procedura:

        1. trasformiamo la query in un embedding;
        2. confrontiamo la query con tutti i documenti;
        3. calcoliamo la similarità coseno;
        4. ordiniamo i risultati;
        5. restituiamo i primi n documenti.
        """

        if documenti is None:

            documenti = self.documenti

        embedding_query = self.documento_to_embedding(
            query
        )

        similarita = cosine_similarity(
            [embedding_query],
            self.matrice_embeddings
        )[0]

        risultati = [

            (
                indice + 1,
                valore_similarita,
                documento
            )

            for indice, (valore_similarita, documento)
            in enumerate(
                zip(
                    similarita,
                    documenti
                )
            )
        ]

        risultati.sort(
            key=lambda elemento: elemento[1],
            reverse=True
        )

        return risultati[:n]


# ============================================================
# 15. TEST DELLA CLASSE PERSONALIZZATA
# ============================================================

print("\n" + "=" * 70)
print("8. TEST DEL SISTEMA PERSONALIZZATO")
print("=" * 70)

mio_sistema = MioSistemaEmbeddings(
    metodo="combinato",
    modello_w2v=model_w2v
)

mio_sistema.addestra(
    docs_test
)

risultati_personalizzati = mio_sistema.trova_simili(
    query_test,
    n=3
)

print("\nQuery utilizzata:")
print(query_test)

print("\nRisultati del sistema personalizzato:")

for posizione, (doc_id, sim, documento) in enumerate(
    risultati_personalizzati,
    1
):

    print(
        f"{posizione}. Doc {doc_id} "
        f"(similarità: {sim:.4f}) - {documento}"
    )


# ============================================================
# 16. CONFRONTO TRA DIVERSI METODI
# ============================================================

"""
Confrontiamo i seguenti metodi:

- media:
  media semplice degli embeddings Word2Vec;

- pesato:
  assegna più importanza ad alcune parole;

- posizione:
  considera anche la posizione delle parole;

- normalizzato:
  considera la lunghezza del documento;

- combinato:
  combina Word2Vec e TF-IDF.
"""

print("\n" + "=" * 70)
print("9. CONFRONTO TRA METODI")
print("=" * 70)

metodi_da_confrontare = [
    "media",
    "pesato",
    "posizione",
    "normalizzato",
    "combinato"
]

risultati_metodi = {}

for metodo in metodi_da_confrontare:

    print(f"\nAddestramento con metodo: {metodo}")

    sistema = MioSistemaEmbeddings(
        metodo=metodo,
        modello_w2v=model_w2v
    )

    sistema.addestra(
        docs_test
    )

    risultati = sistema.trova_simili(
        query_test,
        n=5
    )

    risultati_metodi[metodo] = risultati

    print("Ranking ottenuto:")

    for posizione, (doc_id, sim, documento) in enumerate(
        risultati,
        1
    ):

        print(
            f"{posizione}. Doc {doc_id} "
            f"(similarità: {sim:.4f}) - {documento}"
        )


# ============================================================
# 17. GRAFICO DELLE SIMILARITÀ DEI DOCUMENTI
# ============================================================

"""
Visualizziamo le similarità ottenute dai diversi metodi.

Ogni gruppo di barre rappresenta un documento.

In questo modo possiamo osservare se i metodi producono
risultati simili oppure se modificano il ranking.
"""

print("\n" + "=" * 70)
print("10. GRAFICO DI CONFRONTO DELLE SIMILARITÀ")
print("=" * 70)

indici_documenti = np.arange(
    len(docs_test)
)

larghezza_barra = 0.15

plt.figure(figsize=(14, 7))

for indice_metodo, metodo in enumerate(metodi_da_confrontare):

    risultati = risultati_metodi[metodo]

    valori_similarita = np.zeros(
        len(docs_test)
    )

    for doc_id, sim, documento in risultati:

        valori_similarita[doc_id - 1] = sim

    posizioni = (
        indici_documenti
        +
        indice_metodo * larghezza_barra
    )

    plt.bar(
        posizioni,
        valori_similarita,
        width=larghezza_barra,
        label=metodo
    )

plt.xticks(
    indici_documenti
    +
    larghezza_barra * 2,
    [
        f"Doc {indice + 1}"
        for indice in range(len(docs_test))
    ]
)

plt.title("Confronto delle similarità tra i diversi metodi")
plt.xlabel("Documenti")
plt.ylabel("Similarità coseno")
plt.legend()
plt.grid(axis="y")
plt.show()


# ============================================================
# 18. SFIDA 5:
# FILTRI PER CATEGORIA E DATA
# ============================================================

"""
In un sistema reale i documenti possono avere metadati.

Esempi di metadati:

- categoria;
- data di pubblicazione;
- autore;
- lingua;
- argomento.

Creiamo quindi un DataFrame con il testo e alcuni metadati.
"""

print("\n" + "=" * 70)
print("11. FILTRI PER CATEGORIA E DATA")
print("=" * 70)

documenti_info = [

    {
        "testo": docs_test[0],
        "categoria": "AI",
        "data": "2025-01-10",
        "lingua": "it"
    },

    {
        "testo": docs_test[1],
        "categoria": "Animali",
        "data": "2024-06-15",
        "lingua": "it"
    },

    {
        "testo": docs_test[2],
        "categoria": "Programmazione",
        "data": "2025-03-20",
        "lingua": "it"
    },

    {
        "testo": docs_test[3],
        "categoria": "AI",
        "data": "2025-04-05",
        "lingua": "it"
    },

    {
        "testo": docs_test[4],
        "categoria": "Animali",
        "data": "2023-11-11",
        "lingua": "it"
    }
]

df_documenti = pd.DataFrame(
    documenti_info
)

df_documenti["data"] = pd.to_datetime(
    df_documenti["data"]
)

print("\nArchivio documenti:")
print(df_documenti)


def filtra_documenti(
    archivio,
    categoria=None,
    data_inizio=None,
    data_fine=None,
    lingua=None
):

    """
    Filtra i documenti in base ai metadati.

    I filtri sono opzionali.

    Se un filtro è None, non viene applicato.
    """

    risultato = archivio.copy()

    if categoria is not None:

        risultato = risultato[
            risultato["categoria"] == categoria
        ]

    if data_inizio is not None:

        risultato = risultato[
            risultato["data"] >= pd.to_datetime(data_inizio)
        ]

    if data_fine is not None:

        risultato = risultato[
            risultato["data"] <= pd.to_datetime(data_fine)
        ]

    if lingua is not None:

        risultato = risultato[
            risultato["lingua"] == lingua
        ]

    return risultato


documenti_filtrati = filtra_documenti(
    df_documenti,
    categoria="AI",
    data_inizio="2025-01-01",
    data_fine="2025-12-31"
)

print("\nDocumenti filtrati per categoria AI e anno 2025:")
print(documenti_filtrati)


# ============================================================
# 19. RICERCA DOPO L'APPLICAZIONE DEI FILTRI
# ============================================================

"""
Dopo aver applicato i filtri, passiamo al sistema soltanto
i testi dei documenti rimasti.

In questo modo il sistema non confronta la query con documenti
che non rispettano i criteri scelti dall'utente.
"""

testi_filtrati = documenti_filtrati["testo"].tolist()

if len(testi_filtrati) > 0:

    sistema_filtrato = MioSistemaEmbeddings(
        metodo="combinato",
        modello_w2v=model_w2v
    )

    sistema_filtrato.addestra(
        testi_filtrati
    )

    risultati_filtrati = sistema_filtrato.trova_simili(
        query_test,
        n=3
    )

    print("\nRisultati dopo l'applicazione dei filtri:")

    for posizione, (doc_id, sim, documento) in enumerate(
        risultati_filtrati,
        1
    ):

        print(
            f"{posizione}. "
            f"(similarità: {sim:.4f}) - {documento}"
        )

else:

    print(
        "\nNessun documento rispetta i filtri selezionati."
    )


# ============================================================
# 20. SFIDA 6:
# METRICHE DI VALUTAZIONE
# ============================================================

"""
Per valutare un sistema di ricerca dobbiamo sapere quali
documenti sono realmente rilevanti per una determinata query.

Queste informazioni vengono chiamate ground truth.

Per la query:

"Voglio imparare l'intelligenza artificiale"

ipotizziamo che i documenti rilevanti siano:

- Doc 1: machine learning e intelligenza artificiale;
- Doc 4: deep learning e reti neurali.

Questa etichettatura è didattica e viene definita manualmente.
"""

print("\n" + "=" * 70)
print("12. VALUTAZIONE DEL SISTEMA")
print("=" * 70)

documenti_rilevanti = {
    1,
    4
}


def precision_at_k(
    risultati,
    documenti_rilevanti,
    k
):

    """
    Precision@k:

    misura quanti dei primi k risultati sono rilevanti.

    Formula:

    Precision@k =
        documenti rilevanti nei primi k risultati / k
    """

    primi_k = risultati[:k]

    if len(primi_k) == 0:

        return 0.0

    rilevanti_trovati = sum(

        1

        for doc_id, sim, documento
        in primi_k

        if doc_id in documenti_rilevanti
    )

    return rilevanti_trovati / len(primi_k)


def recall_at_k(
    risultati,
    documenti_rilevanti,
    k
):

    """
    Recall@k:

    misura quanti dei documenti rilevanti totali
    sono stati recuperati nei primi k risultati.

    Formula:

    Recall@k =
        documenti rilevanti recuperati / documenti rilevanti totali
    """

    if len(documenti_rilevanti) == 0:

        return 0.0

    primi_k = risultati[:k]

    rilevanti_trovati = sum(

        1

        for doc_id, sim, documento
        in primi_k

        if doc_id in documenti_rilevanti
    )

    return rilevanti_trovati / len(documenti_rilevanti)


print("\nValutazione dei diversi metodi:")

valutazioni = []

for metodo in metodi_da_confrontare:

    risultati = risultati_metodi[metodo]

    precisione = precision_at_k(
        risultati,
        documenti_rilevanti,
        k=3
    )

    recupero = recall_at_k(
        risultati,
        documenti_rilevanti,
        k=3
    )

    valutazioni.append({

        "metodo": metodo,
        "precision_at_3": precisione,
        "recall_at_3": recupero
    })

    print(
        f"{metodo}: "
        f"Precision@3 = {precisione:.4f}, "
        f"Recall@3 = {recupero:.4f}"
    )

df_valutazioni = pd.DataFrame(
    valutazioni
)

print("\nTabella riassuntiva delle metriche:")
print(df_valutazioni)


# ============================================================
# 21. GRAFICO DELLE METRICHE
# ============================================================

plt.figure(figsize=(10, 6))

x = np.arange(
    len(df_valutazioni["metodo"])
)

larghezza = 0.35

plt.bar(
    x - larghezza / 2,
    df_valutazioni["precision_at_3"],
    width=larghezza,
    label="Precision@3"
)

plt.bar(
    x + larghezza / 2,
    df_valutazioni["recall_at_3"],
    width=larghezza,
    label="Recall@3"
)

plt.xticks(
    x,
    df_valutazioni["metodo"]
)

plt.ylim(0, 1.1)
plt.title("Valutazione dei metodi")
plt.xlabel("Metodo")
plt.ylabel("Valore della metrica")
plt.legend()
plt.grid(axis="y")
plt.show()


# ============================================================
# 22. SFIDA 7:
# SISTEMA DI FEEDBACK
# ============================================================

"""
Un sistema di ricerca può migliorare grazie al feedback dell'utente.

Esempio:

1. il sistema restituisce alcuni documenti;
2. l'utente indica quali risultati sono utili;
3. il sistema aumenta il punteggio dei documenti utili;
4. il sistema diminuisce il punteggio dei documenti non utili.

In questo esercizio realizziamo un feedback semplice.

Non stiamo riaddestrando Word2Vec.

Stiamo modificando il ranking finale dei risultati.
"""


def applica_feedback(
    risultati,
    documenti_positivi=None,
    documenti_negativi=None,
    incremento=0.20,
    decremento=0.20
):

    """
    Modifica i punteggi in base al feedback dell'utente.

    documenti_positivi:
        insieme degli ID giudicati utili.

    documenti_negativi:
        insieme degli ID giudicati non utili.

    incremento:
        quantità aggiunta al punteggio dei documenti positivi.

    decremento:
        quantità sottratta al punteggio dei documenti negativi.
    """

    if documenti_positivi is None:

        documenti_positivi = set()

    if documenti_negativi is None:

        documenti_negativi = set()

    nuovi_risultati = []

    for doc_id, sim, documento in risultati:

        nuovo_punteggio = sim

        if doc_id in documenti_positivi:

            nuovo_punteggio = nuovo_punteggio + incremento

        if doc_id in documenti_negativi:

            nuovo_punteggio = nuovo_punteggio - decremento

        nuovi_risultati.append(

            (
                doc_id,
                nuovo_punteggio,
                documento
            )
        )

    nuovi_risultati.sort(
        key=lambda elemento: elemento[1],
        reverse=True
    )

    return nuovi_risultati


print("\n" + "=" * 70)
print("13. SISTEMA DI FEEDBACK")
print("=" * 70)

risultati_prima_del_feedback = risultati_metodi["combinato"]

print("\nRanking prima del feedback:")

for posizione, (doc_id, sim, documento) in enumerate(
    risultati_prima_del_feedback,
    1
):

    print(
        f"{posizione}. Doc {doc_id} "
        f"(punteggio: {sim:.4f}) - {documento}"
    )

# Supponiamo che l'utente giudichi rilevanti Doc 1 e Doc 4.
feedback_positivi = {
    1,
    4
}

# Supponiamo che l'utente giudichi non rilevante Doc 2.
feedback_negativi = {
    2
}

risultati_dopo_feedback = applica_feedback(
    risultati_prima_del_feedback,
    documenti_positivi=feedback_positivi,
    documenti_negativi=feedback_negativi
)

print("\nRanking dopo il feedback:")

for posizione, (doc_id, sim, documento) in enumerate(
    risultati_dopo_feedback,
    1
):

    print(
        f"{posizione}. Doc {doc_id} "
        f"(punteggio: {sim:.4f}) - {documento}"
    )


# ============================================================
# 23. SFIDA 8:
# GESTIONE DI DOCUMENTI MULTILINGUE
# ============================================================

"""
Aggiungiamo alcuni documenti in inglese.

Ogni documento possiede anche un campo "lingua".

In un sistema reale potremmo:

- utilizzare un modello multilingue;
- utilizzare modelli diversi per ogni lingua;
- tradurre i documenti;
- filtrare i documenti in base alla lingua;
- utilizzare embeddings multilingue pre-addestrati.

Il nostro modello Word2Vec, invece, è stato addestrato
su un corpus principalmente italiano.

Perciò la gestione multilingue qui è soltanto dimostrativa.
"""

print("\n" + "=" * 70)
print("14. GESTIONE MULTILINGUE")
print("=" * 70)

documenti_multilingue = [

    {
        "testo": "Il machine learning impara dai dati",
        "lingua": "it",
        "categoria": "AI"
    },

    {
        "testo": "Machine learning learns from data",
        "lingua": "en",
        "categoria": "AI"
    },

    {
        "testo": "Le reti neurali sono usate nel deep learning",
        "lingua": "it",
        "categoria": "AI"
    },

    {
        "testo": "Neural networks are used in deep learning",
        "lingua": "en",
        "categoria": "AI"
    }
]

df_multilingue = pd.DataFrame(
    documenti_multilingue
)

print("\nArchivio multilingue:")
print(df_multilingue)


# Selezioniamo soltanto i documenti italiani.
documenti_italiani = df_multilingue[
    df_multilingue["lingua"] == "it"
]["testo"].tolist()

print("\nDocumenti italiani:")
for documento in documenti_italiani:
    print("-", documento)


# Selezioniamo soltanto i documenti inglesi.
documenti_inglesi = df_multilingue[
    df_multilingue["lingua"] == "en"
]["testo"].tolist()

print("\nDocumenti inglesi:")
for documento in documenti_inglesi:
    print("-", documento)


# ============================================================
# 24. CONFRONTO TRA DOCUMENTI DI LINGUE DIVERSE
# ============================================================

"""
Il modello Word2Vec utilizzato è stato addestrato su un corpus
italiano molto piccolo.

Le parole inglesi potrebbero non essere presenti nel vocabolario.

Controlliamo quindi quali parole sono conosciute dal modello.
"""

parole_inglesi_da_controllare = [
    "machine",
    "learning",
    "learns",
    "data",
    "neural",
    "networks",
    "deep"
]

print("\nParole inglesi presenti nel vocabolario Word2Vec:")

for parola in parole_inglesi_da_controllare:

    if parola in model_w2v.wv:

        print(f"- {parola}: presente")

    else:

        print(f"- {parola}: non presente")


# ============================================================
# 25. MATRICE DI SIMILARITÀ TRA I DOCUMENTI
# ============================================================

"""
Calcoliamo la similarità tra tutti i documenti.

La matrice risultante contiene:

- righe: documenti;
- colonne: documenti;
- valori: similarità coseno.

La diagonale principale contiene valori vicini a 1,
perché ogni documento è identico a se stesso.
"""

print("\n" + "=" * 70)
print("15. MATRICE DI SIMILARITÀ TRA DOCUMENTI")
print("=" * 70)

matrice_similarita_documenti = cosine_similarity(
    mio_sistema.matrice_embeddings
)

print(
    pd.DataFrame(
        matrice_similarita_documenti,
        index=[
            f"Doc {i + 1}"
            for i in range(len(docs_test))
        ],
        columns=[
            f"Doc {i + 1}"
            for i in range(len(docs_test))
        ]
    )
)

plt.figure(figsize=(8, 6))

plt.imshow(
    matrice_similarita_documenti
)

plt.colorbar(
    label="Similarità coseno"
)

plt.xticks(
    range(len(docs_test)),
    [
        f"Doc {i + 1}"
        for i in range(len(docs_test))
    ]
)

plt.yticks(
    range(len(docs_test)),
    [
        f"Doc {i + 1}"
        for i in range(len(docs_test))
    ]
)

plt.title("Matrice di similarità tra i documenti")
plt.xlabel("Documenti")
plt.ylabel("Documenti")
plt.show()


# ============================================================
# 26. PCA DEGLI EMBEDDINGS DEI DOCUMENTI
# ============================================================

"""
Anche gli embeddings dei documenti possono essere visualizzati
in uno spazio a due dimensioni.

Utilizziamo PCA per ridurre i vettori dei documenti
a due componenti principali.
"""

if len(docs_test) >= 2:

    pca_documenti = PCA(
        n_components=2
    )

    coordinate_documenti = pca_documenti.fit_transform(
        mio_sistema.matrice_embeddings
    )

    plt.figure(figsize=(10, 7))

    plt.scatter(
        coordinate_documenti[:, 0],
        coordinate_documenti[:, 1]
    )

    for indice, documento in enumerate(docs_test):

        plt.annotate(
            f"Doc {indice + 1}",
            (
                coordinate_documenti[indice, 0],
                coordinate_documenti[indice, 1]
            )
        )

    plt.title(
        "Visualizzazione degli embeddings dei documenti con PCA"
    )

    plt.xlabel("Componente principale 1")
    plt.ylabel("Componente principale 2")
    plt.grid(True)
    plt.show()


# ============================================================
# 27. ANALISI FINALE DEI RISULTATI
# ============================================================

print("\n" + "=" * 70)
print("16. ANALISI FINALE")
print("=" * 70)

print(
    """
CONCLUSIONI:

1. TF-IDF rappresenta i documenti usando la frequenza delle parole.

2. Word2Vec rappresenta le parole attraverso vettori densi
   appresi osservando i contesti.

3. Per rappresentare un documento con Word2Vec possiamo calcolare
   la media degli embeddings delle parole che lo compongono.

4. L'utilizzo di pesi diversi permette di dare maggiore importanza
   ad alcune parole considerate significative.

5. L'informazione sulla posizione può essere aggiunta modificando
   il peso delle parole in base alla loro posizione nel documento.

6. La normalizzazione aiuta a ridurre l'effetto della lunghezza
   dei documenti.

7. TF-IDF e Word2Vec possono essere combinati per ottenere
   una rappresentazione che utilizza sia informazione lessicale
   sia informazione semantica.

8. I filtri per categoria, data e lingua permettono di restringere
   l'insieme dei documenti analizzati.

9. Precision@k e Recall@k permettono di valutare la qualità
   dei risultati, se possediamo una ground truth.

10. Il feedback dell'utente può modificare il ranking dei documenti.

11. Per gestire davvero più lingue sarebbe preferibile utilizzare
    embeddings multilingue pre-addestrati.

12. Un corpus piccolo produce risultati didattici ma non sempre
    semanticamente affidabili.
"""
)


# ============================================================
# 28. CHECKLIST DELLE SFIDE COMPLETATE
# ============================================================

print("\n" + "=" * 70)
print("17. CHECKLIST DELLE SFIDE")
print("=" * 70)

sfide_completate = [

    "Combinazione di TF-IDF e Word2Vec",
    "Pesi diversi per parole importanti",
    "Utilizzo della posizione delle parole",
    "Normalizzazione per la lunghezza del documento",
    "Filtri per categoria, data e lingua",
    "Metriche Precision@k e Recall@k",
    "Sistema di feedback sui risultati",
    "Gestione dimostrativa di documenti multilingue"
]

for indice, sfida in enumerate(sfide_completate, 1):

    print(
        f"{indice}. COMPLETATA - {sfida}"
    )


# ============================================================
# 29. POSSIBILI ESTENSIONI
# ============================================================

print("\n" + "=" * 70)
print("18. POSSIBILI ESTENSIONI")
print("=" * 70)

print(
    """
PER CONTINUARE L'ESERCIZIO:

- utilizzare un corpus più grande;
- provare embeddings Word2Vec pre-addestrati;
- confrontare Word2Vec con FastText;
- utilizzare GloVe;
- provare embeddings multilingue;
- confrontare la media semplice con la media pesata;
- utilizzare pesi TF-IDF direttamente sugli embeddings Word2Vec;
- aggiungere stemming o lemmatizzazione;
- eliminare le stopword;
- introdurre una vera fase di riaddestramento basata sul feedback;
- confrontare i risultati con un modello BERT;
- misurare il tempo di esecuzione;
- creare una piccola applicazione di ricerca semantica.
"""
)
```
