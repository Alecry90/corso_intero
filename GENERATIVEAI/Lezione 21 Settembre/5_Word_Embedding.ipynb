```python
import gensim.downloader as api
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
# Comment here...

model = api.load("word2vec-google-news-300")

# Comment here...
```

```python
word = 'king'
similar_words = model.most_similar(word, topn=10)
print(similar_words)
# Comment here...
```

```python
analogy = model.most_similar(positive=['king', 'woman'], negative=['man'], topn=1)
print(analogy)
# Comment here...
```

```python
word1 = 'king'
word2 = 'queen'
similarity = model.similarity(word1, word2)
print(similarity)
```

```python
from gensim.models import Word2Vec

sentences = [['this', 'is', 'the', 'first', 'sentence', 'for', 'word2vec'],
             ['this', 'is', 'the', 'second', 'sentence'],
             ['yet', 'another', 'sentence'],
             ['one', 'more', 'sentence'],
             ['and', 'the', 'final', 'sentence']]


#vector_size:	100	Dimensione dello spazio vettoriale. Ogni parola verrà rappresentata da un elenco di 100 numeri.
#window:	5	Dimensione della finestra di contesto. Il modello guarda fino a 5 parole prima e 5 parole dopo la parola target per capire il suo significato.
#min_count:	1	Soglia di frequenza minima. Ignora le parole che compaiono meno di 1 volta (impostato a 1, include tutte le parole presenti).
#sg:	0	Algoritmo di addestramento. 0 indica CBOW (Continuous Bag of Words, predice la parola target dal contesto), mentre 1 userebbe Skip-gram (predice il contesto dalla parola).

model = Word2Vec(sentences, vector_size=100, window=5, min_count=1, sg=0)
```

```python
words = list(model.wv.index_to_key)
X = [model.wv[word] for word in words]


pca = PCA(n_components=2)
X_2d = pca.fit_transform(X)

plt.scatter(X_2d[:,0], X_2d[:,1])
for i, word in enumerate(words):
    plt.annotate(word, (X_2d[i,0], X_2d[i,1]))
plt.show()
```
