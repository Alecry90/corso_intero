***Domanda 1 — Scelta della soluzione di Intelligenza Artificiale***
Un'organizzazione deve scegliere quale soluzione di Intelligenza Artificiale adottare in due contesti con caratteristiche differenti.

Contesto 1 — Assicurazione: La soluzione deve supportare un operatore nella valutazione del livello di rischio associato a una richiesta di assicurazione.

Contesto 2 — Selezione di contenuti online: La soluzione deve selezionare e proporre agli utenti contenuti online, come articoli, video o post, sulla base delle loro caratteristiche e delle loro precedenti interazioni.

Per entrambi i contesti sono disponibili due alternative:

La soluzione A che ha raggiunto durante i test un'accuratezza nelle previsioni del 94% e si basa su una rete neurale complessa, caratterizzata da numerosi livelli e parametri, poco interpretabile.

La soluzione B che ha raggiunto durante i test un'accuratezza nelle previsioni dell'89% e si basa su un albero decisionale, nel quale le decisioni vengono effettuate attraverso una sequenza di condizioni.

Per ciascun contesto, analizza i vantaggi e gli svantaggi delle due soluzioni, considerando le caratteristiche del processo decisionale, i soggetti coinvolti e le possibili conseguenze derivanti dall'utilizzo delle previsioni prodotte.

✏️ Risposta:


Nel contesto assicurativo sceglierei la **soluzione B**, almeno come supporto all'operatore. In questo caso la previsione può influire sulle condizioni economiche offerte al cliente, quindi è importante poter capire quali elementi hanno portato alla valutazione del rischio. Un albero decisionale rende più leggibili i passaggi della decisione rispetto a una rete neurale complessa e può aiutare l'operatore a controllare e motivare il suggerimento. Non garantisce però da solo una spiegazione completa o l'assenza di errori e discriminazioni: bisogna verificare anche le variabili usate e il comportamento del modello. La soluzione A ha un'accuratezza maggiore, ma il 94% contro l'89% non basta per preferirla senza sapere quali errori commettono i modelli e quanto siano gravi. Perciò, in questo caso, darei più peso alla trasparenza, mantenendo comunque la supervisione umana.

Per la selezione dei contenuti online sceglierei invece la **soluzione A**, perché una rete neurale può cogliere relazioni complesse tra caratteristiche e interazioni degli utenti. La scelta è però valida solo se il 94% è stato misurato con una metrica adatta allo scopo: l'accuratezza, da sola, non dimostra che i contenuti siano più pertinenti o che gli utenti siano più soddisfatti. Considererei anche la varietà dei suggerimenti e il rischio di creare bolle informative; quindi non definirei l'impatto sempre trascurabile. La soluzione B è più semplice da interpretare e modificare, ma potrebbe rappresentare peggio comportamenti e preferenze che cambiano nel tempo.

In entrambi i casi la scelta dipende dal contesto, dalle conseguenze degli errori e da come viene verificato il modello, non solo dalla percentuale di accuratezza. Questo ragionamento è coerente con i concetti generali di reti neurali e valutazione affrontati a lezione; alberi decisionali, sistemi di raccomandazione e norme come GDPR e AI Act non risultano invece approfonditi direttamente nei materiali consultati, quindi li considero un'applicazione dei concetti al caso proposto.

***Domanda 2 — Analisi di un sistema di Intelligenza Artificiale***

Un'azienda utilizza un sistema di Intelligenza Artificiale per trascrivere automaticamente le conversazioni degli utenti. Nei test il sistema raggiunge un'accuratezza complessiva del 94%, ma il team osserva che la percentuale di errori varia tra persone con caratteristiche differenti.

Analizza il caso indicando da dove potrebbe avere origine il problema e quali interventi adotteresti per ridurlo.

✏️ **Risposta:**

Il 94% di accuratezza complessiva non basta, da solo, per stabilire se il sistema sia adeguato: la media potrebbe nascondere prestazioni peggiori per alcuni utenti. Prima di concludere che ci sia una discriminazione, verificherei quali gruppi e condizioni presentano più errori, confrontando le trascrizioni con quelle corrette e misurando gli errori separatamente. Le cause potrebbero essere dati di addestramento poco rappresentativi, differenze di accento o lingua, rumori e qualità audio, oppure trascrizioni di riferimento non corrette.

Per ridurre il problema raccoglierei, rispettando la privacy, esempi audio più vari e rappresentativi; controllerei inoltre la qualità delle trascrizioni usate per addestrare il sistema. Dopo aver aggiornato il modello, lo testerei su dati separati e confronterei sia l'accuratezza generale sia gli errori dei diversi gruppi. Continuerei a monitorarlo anche dopo l'introduzione e prevederei il controllo umano nei casi in cui la trascrizione è incerta. In questo modo valuterei il sistema in base all'uso concreto e non soltanto al risultato medio.
