## Esercizio 1 — Classi, `__init__`, metodi, ereditarietà (`super()`)

Crea una classe `ContoBancario` con:
- `__init__(self, titolare, saldo=0)` che salva `titolare` e `saldo`
- `deposita(self, importo)`: aumenta il saldo. Usa `assert importo > 0, "l'importo deve essere positivo"`
- `preleva(self, importo)`: se `importo <= saldo` diminuisce il saldo, altrimenti stampa un messaggio di errore (senza sollevare eccezioni)
- `mostra_saldo(self)`: stampa `titolare` e `saldo` formattato con **2 decimali** (f-string)

Poi crea una sottoclasse `ContoRisparmio(ContoBancario)` che:
- nel costruttore chiama `super().__init__(titolare, saldo)` e salva anche `tasso_interesse` (es. 0.02)
- ha un metodo `applica_interessi(self)` che aumenta il saldo di `saldo * tasso_interesse`

Testa creando un `ContoRisparmio`, depositando, prelevando (anche un importo troppo alto) e applicando gli interessi.

---
### 📎 Teoria & dove trovarla nel notebook originale

Una **classe** è un modello per creare oggetti: `__init__` (il *costruttore*) viene eseguito alla creazione dell'oggetto e salva i dati iniziali su `self` (l'oggetto stesso); gli altri metodi sono funzioni che operano su quei dati. Una classe `Figlia(Genitore)` **eredita** attributi e metodi del genitore; dentro `__init__` della figlia, `super().__init__(...)` chiama il costruttore del genitore invece di riscriverlo da capo.

Nel notebook annotato:
- classe con `__init__`, metodi e docstring → `class SelfAttention:` (**Blocco 3**, riga 616)
- ereditarietà + `super()` → `class PositionalEncoding(layers.Layer): def __init__(...): super().__init__()` (**Blocco 5**, righe 1096-1105)

````python
class ContoBancario:
    """Rappresenta un semplice conto bancario."""

    def __init__(self, titolare, saldo=0):
        # TODO: salva titolare e saldo come attributi
        self.titolare= titolare
        self.saldo= saldo


    def deposita(self, importo):
        # TODO: assert + aggiornamento saldo
            assert importo > 0
            self.saldo += importo

    def preleva(self, importo):
        # TODO: se importo <= saldo diminuisci il saldo, altrimenti stampa un errore
        if importo < 0:
            self.saldo -= importo

    def mostra_saldo(self):
        # TODO: stampa titolare e saldo con 2 decimali (f-string)
        print(f"il Saldo del Sg. {self.titolare} è {self.saldo:.2f} ")
        


class ContoRisparmio(ContoBancario):
    def __init__(self, titolare, saldo=0, tasso_interesse=0.02):
        # TODO: chiama super().__init__ e salva tasso_interesse
        super().__init__(titolare, saldo)
        self.tasso_interesse =tasso_interesse

    def applica_interessi(self):
        # TODO: aumenta il saldo in base al tasso di interesse
        interessi= self.saldo*self.tasso_interesse
        self.saldo += interessi
        print(f"Interessi applicati: +{interessi:.2f} EUR")



# --- Test ---
conto = ContoRisparmio("Marco", 1000)
conto.deposita(200)
conto.preleva(5000)
conto.applica_interessi()
conto.mostra_saldo()
````
