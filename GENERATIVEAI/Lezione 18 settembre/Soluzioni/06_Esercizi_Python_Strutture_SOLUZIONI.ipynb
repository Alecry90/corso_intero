## Esercizio 1 — Classi, `__init__`, metodi, ereditarietà (`super()`)

---
### 📎 Teoria & dove trovarla nel notebook originale

Una **classe** è un modello per creare oggetti: `__init__` (il *costruttore*) viene eseguito alla creazione dell'oggetto e salva i dati iniziali su `self` (l'oggetto stesso); gli altri metodi sono funzioni che operano su quei dati. Una classe `Figlia(Genitore)` **eredita** attributi e metodi del genitore; dentro `__init__` della figlia, `super().__init__(...)` chiama il costruttore del genitore invece di riscriverlo da capo.

Nel notebook annotato:
- classe con `__init__`, metodi e docstring → `class SelfAttention:` (**Blocco 3**, riga 616)
- ereditarietà + `super()` → `class PositionalEncoding(layers.Layer): def __init__(...): super().__init__()` (**Blocco 5**, righe 1096-1105)

```python
class ContoBancario:
    """Rappresenta un semplice conto bancario."""

    def __init__(self, titolare, saldo=0):
        self.titolare = titolare
        self.saldo = saldo

    def deposita(self, importo):
        assert importo > 0, "l'importo deve essere positivo"
        self.saldo += importo

    def preleva(self, importo):
        if importo <= self.saldo:
            self.saldo -= importo
        else:
            print(f"Errore: saldo insufficiente per prelevare {importo:.2f}")

    def mostra_saldo(self):
        print(f"{self.titolare}: saldo = {self.saldo:.2f} EUR")


class ContoRisparmio(ContoBancario):
    def __init__(self, titolare, saldo=0, tasso_interesse=0.02):
        super().__init__(titolare, saldo)
        self.tasso_interesse = tasso_interesse

    def applica_interessi(self):
        interessi = self.saldo * self.tasso_interesse
        self.saldo += interessi
        print(f"Interessi applicati: +{interessi:.2f} EUR")


# --- Test ---
conto = ContoRisparmio("Marco", 1000)
conto.deposita(200)
conto.preleva(5000)   # saldo insufficiente -> messaggio di errore
conto.applica_interessi()
conto.mostra_saldo()
```

## Esercizio 2 — Funzioni: parametri di default, return multipli, ricorsione

---
### 📎 Teoria & dove trovarla nel notebook originale

Un parametro di default (`def f(x, y=10):`) permette di chiamare la funzione anche omettendo quell'argomento. Una funzione può restituire più valori insieme separandoli con la virgola dopo `return`: Python li impacchetta in una **tupla**, che si può "spacchettare" (unpacking) alla chiamata (`a, b, c = f(...)`). Una funzione **ricorsiva** chiama se stessa su un problema più piccolo, fino a un **caso base** che interrompe la catena di chiamate.

Nel notebook annotato:
- parametro di default → `def translation(self, text, target_lang="italian"):` (**Blocco 7**, riga 1756)
- return multiplo + unpacking → `return output, attention_weights` (**Blocco 3**, righe 675/702) e `output, attention_weights = attention.forward(X)` (riga 730)
- ricorsione → il notebook la mostra solo come **testo di esempio** (non eseguito) dentro `code_generation`, con `fibonacci`/`factorial` (**Blocco 7**, righe 1803-1814): qui la farai girare davvero.

```python
def statistiche(numeri):
    minimo = min(numeri)
    massimo = max(numeri)
    media = sum(numeri) / len(numeri)
    return minimo, massimo, media


def saluta(nome, lingua="it"):
    if lingua == "it":
        return f"Ciao, {nome}!"
    elif lingua == "en":
        return f"Hello, {nome}!"
    else:
        return f"({lingua}) Hi, {nome}!"


def fattoriale(n):
    if n <= 1:
        return 1
    return n * fattoriale(n - 1)


def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)


# --- Test ---
minimo, massimo, media = statistiche([4, 8, 15, 16, 23, 42])
print(minimo, massimo, media)
print(saluta("Anna"))
print(saluta("Anna", lingua="en"))
print(fattoriale(5))
print([fibonacci(i) for i in range(8)])
```

## Esercizio 3 — Cicli `for`, `range`, cicli annidati, `enumerate`

---
### 📎 Teoria & dove trovarla nel notebook originale

Un ciclo `for elemento in range(n):` ripete un blocco di codice una volta per ogni valore prodotto da `range`. Due `for` uno dentro l'altro (**ciclo annidato**) ripetono per intero il ciclo interno a ogni iterazione di quello esterno. `enumerate(lista)` restituisce coppie `(indice, elemento)`; `enumerate(lista, k)` fa partire l'indice da `k` invece che da 0.

Nel notebook annotato:
- ciclo annidato con `enumerate` → **Blocco 2**, righe 447-449: `for i, (x1, y1) in enumerate(positions): for j, (x2, y2) in enumerate(positions):`
- `enumerate` con partenza da 1 → **Blocco 6**, riga 1492: `for i, prompt in enumerate(prompts, 1):`

```python
# a) tabellina 1-5 con doppio ciclo for
for i in range(1, 6):
    for j in range(1, 6):
        print(f"{i} x {j} = {i * j}")
    print()  # riga vuota tra un blocco e l'altro


# b) elenco numerato con enumerate(..., 1)
prodotti = ["pane", "latte", "uova", "mele"]
for indice, prodotto in enumerate(prodotti, 1):
    print(f"{indice}. {prodotto}")
```

## Esercizio 4 — List comprehension (anche con espressione ternaria dentro)

---
### 📎 Teoria & dove trovarla nel notebook originale

Una **list comprehension** `[espressione for elemento in lista]` costruisce una nuova lista applicando `espressione` a ogni elemento, in una riga. Si può filtrare con `[espressione for elemento in lista if condizione]`, oppure scegliere tra due valori diversi con un **operatore ternario** dentro l'espressione: `[a if condizione else b for elemento in lista]`.

Nel notebook annotato è praticamente lo stesso pattern: **Blocco 6**, riga 1617 — `colors = ['red' if i in top_k_indices else 'lightgray' for i in range(vocab_size)]`.

```python
voti = [4, 6, 8, 5, 9, 3, 7]

# a) voti raddoppiati
raddoppiati = [v * 2 for v in voti]

# b) solo voti sufficienti
sufficienti = [v for v in voti if v >= 6]

# c) etichette con ternario dentro la comprehension
etichette = ["sufficiente" if v >= 6 else "insufficiente" for v in voti]

print(raddoppiati)
print(sufficienti)
print(etichette)
```

## Esercizio 5 — Dizionari: creazione, `.items()`, lookup

---
### 📎 Teoria & dove trovarla nel notebook originale

Un **dizionario** `{chiave: valore, ...}` associa ogni chiave a un valore. `dizionario.items()` restituisce coppie `(chiave, valore)`, usabili in un `for chiave, valore in ...items():`. `dizionario.get(chiave)` cerca la chiave e restituisce `None` (invece di dare errore) se non esiste — utile per un lookup "sicuro".

Nel notebook annotato: la classe `LLMApplications` (**Blocco 7**) fa lookup in dizionari con `.items()`, es. righe 1737, 1791, 1819 — `for key, completion in self.completions.items():`.

```python
# a) dizionario magazzino
magazzino = {"pane": 20, "latte": 15, "uova": 30, "mele": 12}

# b) report con .items()
for chiave, valore in magazzino.items():
    print(f"{chiave}: {valore} pezzi")


def cerca_prodotto(magazzino, nome):
    return magazzino.get(nome)  # None se la chiave non esiste


print(cerca_prodotto(magazzino, "latte"))
print(cerca_prodotto(magazzino, "cioccolato"))

# d) aggiungi/aggiorna
magazzino["cioccolato"] = 8       # nuovo prodotto
magazzino["pane"] = magazzino["pane"] - 5   # aggiornamento quantità
print(magazzino)
```

## Esercizio 6 — Tuple, unpacking, `zip`

---
### 📎 Teoria & dove trovarla nel notebook originale

Una **tupla** `(a, b)` raggruppa più valori in un solo oggetto; si può "spacchettare" direttamente in variabili separate (`x, y = (a, b)`). `zip(lista1, lista2)` accoppia gli elementi di due (o più) liste posizione per posizione, permettendo di scorrerle insieme in un solo `for`.

Nel notebook annotato:
- `zip` + unpacking annidato → **Blocco 2**, riga 439: `for i, ((x, y), label) in enumerate(zip(positions, labels)):`
- versione più semplice → **Blocco 6**, riga 1605: `for temp, color in zip(temperatures, colors):`

```python
nomi = ["Anna", "Luca", "Sara"]
voti = [7, 9, 6]

# a) zip + unpacking nel for
for nome, voto in zip(nomi, voti):
    print(f"{nome}: {voto}")


# b) lista di tuple + ricerca del massimo con un ciclo for
coppie = list(zip(nomi, voti))

migliore = coppie[0]
for nome, voto in coppie:
    if voto > migliore[1]:
        migliore = (nome, voto)

print(coppie)
print(f"Il voto più alto è di {migliore[0]} con {migliore[1]}")
```

## Esercizio 8 — `if`/`elif`/`else`, operatore ternario, `assert`

---
### 📎 Teoria & dove trovarla nel notebook originale

`if/elif/else` esegue un ramo di codice diverso in base a una condizione, valutata in ordine. L'**operatore ternario** `a if condizione else b` è un `if/else` compresso in una sola espressione, comodo quando serve solo scegliere un valore. `assert condizione, "messaggio"` interrompe il programma con un errore se la condizione è falsa: si usa per validare ipotesi che devono sempre essere vere.

Nel notebook annotato:
- `assert` → **Blocco 4**, riga 876: `assert d_model % num_heads == 0`
- operatore ternario, anche annidato → **Blocco 3**, riga 812: `freccia = "↑" if diff > 0 else ("↓" if diff < 0 else "=")`

```python
def classifica_voto(voto):
    assert 0 <= voto <= 10, "voto non valido"
    if voto < 6:
        return "insufficiente"
    elif voto < 7:
        return "sufficiente"
    elif voto < 9:
        return "buono"
    else:
        return "ottimo"


def pari_o_dispari(n):
    return "pari" if n % 2 == 0 else "dispari"


# --- Test ---
for v in [3, 6, 8, 10]:
    print(v, "->", classifica_voto(v))
print(pari_o_dispari(7))
```

## Esercizio 9 — Gestione errori con `try`/`except`

---
### 📎 Teoria & dove trovarla nel notebook originale

Un blocco `try:` esegue codice che potrebbe generare un errore; se l'errore previsto si verifica, l'esecuzione salta al corrispondente `except TipoErrore:` invece di far crashare il programma. Specificare il tipo di eccezione (es. `ZeroDivisionError`, `ValueError`) intercetta solo quell'errore, lasciando propagare gli altri.

Nel notebook annotato: `try/except Exception as e:` attorno al caricamento/uso del modello GPT-2 — **Blocco 6**, righe 1523 e 1559.

```python
def divisione_sicura(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Errore: non è possibile dividere per zero")
        return None


def a_intero_sicuro(testo):
    try:
        return int(testo)
    except ValueError:
        print(f"Errore: '{testo}' non è un numero intero valido")
        return None


# --- Test ---
print(divisione_sicura(10, 2))
print(divisione_sicura(10, 0))
print(a_intero_sicuro("42"))
print(a_intero_sicuro("abc"))
```

## Esercizio 10 — NumPy: array, slicing, `axis`, `argmax`, `dot`, `where`

---
### 📎 Teoria & dove trovarla nel notebook originale

Un array NumPy multidimensionale si indicizza con `array[riga, colonna]`; `array[:, i]` prende **tutta** una colonna, `array[i, :]` **tutta** una riga. Le funzioni di aggregazione (`np.sum`, `np.max`, `np.argmax`, ...) accettano un parametro `axis`: `axis=0` opera lungo le righe (un risultato per colonna), `axis=1` lungo le colonne (un risultato per riga). `np.dot(matrice, vettore)` calcola il prodotto matrice-vettore; `np.where(condizione, a, b)` costruisce un nuovo array scegliendo `a` o `b` elemento per elemento.

Tutti questi costrutti sono usati nella classe `SelfAttention` (**Blocco 3**):
- `np.dot(X, self.W_q)` (righe 696-698)
- `np.max(x, axis=1, keepdims=True)` (riga 685)
- `np.where(mask == 0, -1e9, scores)` (riga 665)
- `np.argmax(attention_weights[parola_focus])` (riga 781)

```python
import numpy as np

vendite = np.array([
    [100, 120, 90],   # negozio 0
    [80,  95,  70],   # negozio 1
    [150, 140, 160],  # negozio 2
    [60,  75,  65],   # negozio 3
])

# a) slicing: secondo mese (colonna) e secondo negozio (riga)
vendite_mese_2 = vendite[:, 1]
vendite_negozio_2 = vendite[1, :]
print("Vendite mese 2 (tutti i negozi):", vendite_mese_2)
print("Vendite negozio 2 (tutti i mesi):", vendite_negozio_2)

# b) totale vendite per negozio e per mese
totale_per_negozio = np.sum(vendite, axis=1)   # somma lungo le colonne -> 1 valore per riga
totale_per_mese = np.sum(vendite, axis=0)      # somma lungo le righe -> 1 valore per colonna
print("Totale per negozio:", totale_per_negozio)
print("Totale per mese:", totale_per_mese)

# c) negozio con le vendite totali più alte
negozio_top = np.argmax(totale_per_negozio)
print(f"Il negozio con più vendite è il negozio {negozio_top} (totale {totale_per_negozio[negozio_top]})")

# d) fatturato pesato con np.dot (peso diverso per ogni mese)
pesi = np.array([1.0, 1.2, 0.9])
fatturato_pesato = np.dot(vendite, pesi)
print("Fatturato pesato per negozio:", fatturato_pesato)

# e) etichette alto/basso con np.where
soglia = 100
etichette = np.where(vendite >= soglia, "alto", "basso")
print(etichette)
```
