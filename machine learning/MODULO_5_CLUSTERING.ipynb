```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.spatial.distance import cdist, pdist, squareform
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, silhouette_samples
from sklearn.datasets import make_blobs, make_moons
np.random.seed(42)

plt.rcParams['figure.figsize']= (7, 4.5)
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3

print("Setup completato.")
```

```python
X, y_true= make_blobs(n_samples= 300, centers=3, cluster_std= 3.80, random_state=42)
plt.figure(figsize=(6, 5))
plt.scatter(X[:, 0], X[:, 1], c=y_true, cmap="viridis", s=25, alpha=0.7)
plt.title("Dati reali")
plt.show()
print(X)
```

```python
#Applicazione di K-Means
#istanziamo 3 cluster(K)
km= KMeans(n_clusters=3, init="k-means++", n_init= 10, random_state=42)
km.fit(X)
```

```python
print("Inertia (WCSS):", round(km.inertia_, 2))
print("Iterazioni effettive:", km.n_iter_)
print("Centroidi:\n",  km.cluster_centers_ , round(2))
```

```python
plt.figure(figsize=(6, 5))
plt.scatter(X[:, 0], X[:, 1], c=km.labels_, cmap="viridis", s=25, alpha=0.7)
plt.scatter(km.cluster_centers_[:, 0], km.cluster_centers_[:,1],
            c="red", marker="X", s=200, edgecolors="white", linewidths=1.5)
plt.title("Dati reali")
plt.show()
print(X)
```

```python
X4, y4_true= make_moons(n_samples= 400, noise= 0.06, random_state= 42)
X4
#Applicazione di K-Means
#istanziamo 2 cluster(K)
km= KMeans(n_clusters=2, init="k-means++", n_init= 10, random_state=42)
km.fit(X4)


plt.figure(figsize=(6, 5))
plt.scatter(X4[:, 0], X4[:, 1], c=km.labels_, cmap="viridis", s=25, alpha=0.7)
plt.scatter(km.cluster_centers_[:, 0], km.cluster_centers_[:,1],
            c="red", marker="X", s=200, edgecolors="white", linewidths=1.5)
plt.title("Dati reali")
plt.show()
```

```python
np.random.seed(42)
n= 200
reddito = np.concatenate([np.random.normal(25,6,50), np.random.normal(55,10,60), np.random.normal(85,9,40),
                           np.random.normal(45,8,50)])
spesa = np.concatenate([np.random.normal(20,8,50), np.random.normal(50,10,60), np.random.normal(85,8,40),
                        np.random.normal(75,10,50)])
df1 = pd.DataFrame({"reddito_kEUR": reddito.clip(min=0), "punteggio_spesa": spesa.clip(0,100)})

X1_scaled= StandardScaler().fit_transform(df1)

x= range(2,10)
lista= []
for k in x:
    km2= KMeans(n_clusters= k, init="k-means++", n_init= 10, random_state=42)
    km2.fit(X1_scaled)
    lista.append(km2.inertia_)

plt.figure(figsize=(6, 5))
plt.plot(x, lista, marker='o', linestyle="--",color= 'blue', linewidth =2 )
plt.grid(True, alpha= 0.3)
plt.title("Dati reali")
plt.xlabel("n°cluster")
plt.ylabel("Wc inertia")
plt.show()


    
    
```
